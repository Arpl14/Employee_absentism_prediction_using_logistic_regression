#### One hot encoding for categorical variables 

```python
# Creating a dummy variables (one-hot encoding) for 'Reason for Absence'
# Each unique reason will get its own column with binary values (0 or 1)
reason_columns = pd.get_dummies(df['Reason for Absence']).astype(int)
reason_columns

```

```python
# Add a 'check' column to validate if each row has only one active reason
# This should ideally be 1 for each row (i.e., only one reason per absence)
reason_columns['check'] = reason_columns.sum(axis=1)
reason_columns
```

```python
# validate that no row has more than one reason
reason_columns['check'].unique()
```

```python
# Dropping the 'check' column after validation
reason_columns = reason_columns.drop(['check'], axis=1)

# Recreating the dummy variables, but this time with `drop_first=True`
# This avoids multicollinearity by dropping one category
reason_columns = pd.get_dummies(df['Reason for Absence'], drop_first=True).astype(int)
reason_columns
```

```python
# Checking the column names in the original df and reason_columns (just for reference)
df.columns.values
```

```python
reason_columns.columns.values
```

if we add reasons_column into the df .. we will again face multicolinairity thus we drop the reason for absense table from df 

```python

# Dropping the original 'Reason for Absence' column from the main dataframe
# since we have already encoded this info using dummy variables
df = df.drop(['Reason for Absence'], axis=1)
```

we have a table from client which explains the reasons for absense :

```python

# Now I want to group the dummy variables (reasons) into broader categories
# based on their ranges — each group represents a type of reason for absence

# Reason Type 1 includes reasons 1 to 14 (serious health issues )
reason_type_1 = reason_columns.loc[:, 1:14].max(axis=1)

# Reason Type 2 includes reasons 15 to 17 (pregnancy related)
reason_type_2 = reason_columns.loc[:, 15:17].max(axis=1)

# Reason Type 3 includes reasons 18 to 21 (food poisioning or other poisioning reasons )
reason_type_3 = reason_columns.loc[:, 18:21].max(axis=1)

# Reason Type 4 includes all other reasons starting from 22 onwards ( general common reasons like dental appointment, headaches etc )
reason_type_4 = reason_columns.loc[:, 22:].max(axis=1)
```

```python
#adding newly created reason for absense classes into our original df 
df = pd.concat([df, reason_type_1, reason_type_2, reason_type_3, reason_type_4], axis = 1)
df
```

```python
column_names = ['Date', 'Transportation Expense', 'Distance to Work', 'Age',
       'Daily Work Load Average', 'Body Mass Index', 'Education',
       'Children', 'Pets', 'Absenteeism Time in Hours', 'Reason_1', 'Reason_2', 'Reason_3', 'Reason_4']
df.columns = column_names
df.head(5)
```

#### Data Preprocessing Summary

In this section, we preprocessed the data by performing the following steps:

- Dropped irrelevant columns (e.g., `ID`) to eliminate noise from the data.
- One-hot encoded the `Reason for Absence` column using `pd.get_dummies()` to convert categorical data into numerical format.
- Grouped reasons into 4 types (`Reason_1` to `Reason_4`) based on column index ranges.
- Concatenated the new reason type columns back into the main DataFrame.
- Mapped `Education` levels to binary (0 = basic, 1 = higher education) for simplification and eliminating class imbalance.
- Reordered and renamed columns for better readability and model compatibility.
- Created a final preprocessed DataFrame ready for analysis or machine learning modeling.
